```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.preprocessing import OneHotEncoder,PolynomialFeatures
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
sns.set_style(style='whitegrid') 
```

```python
##step2: Loading (autompg dataset) and cleaning the data
data=fetch_ucirepo(id=9)
data
```

```python
x=data.data.features.copy()# it returns all numerical values
x
```

```python
y=data.data.targets['mpg']
y

```

```python
x.info()
```

```python
x['origin'].unique();
x['horsepower'].unique()
```

```python
# first we will transform the horsepower and origin columns'
x['horsepower']=pd.to_numeric(x['horsepower'],errors='coerce')
#x['horsepower'].info()
x['origin']=x['origin'].astype(str)
```

```python
#x.info()
#all numerical columns into a single collections
numerical_features=['cylinders','displacement','horsepower','weight','acceleration','model_year']
categorical_features=['origin']
x=x[numerical_features+categorical_features]
x.info()
```

```python
x.isnull().sum()
```

```python
#step3 : Quick exploratory Data Analysis(seaborn)
df=x.copy()
df['mpg']=y
```

```python
#step4 : dividing the data into train test split
print(x)
print(y)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print('\nTrain Test Splint')
print("Training Data",len(x_train))
print("Testing Data",len(x_test))
```

```python
#step5 : creating a baseline model
def evaluate(model_name,actual,predicted):
    """Returns MAE,MSE,RMSE,R2-Score"""
    mse=mean_squared_error(actual,predicted)
    mae=mean_absolute_error(actual,predicted)
    return{'Model_Name':model_name,
    'MAE':mae,
    'MSE':mse,
    'RMSE':np.sqrt(mse),
    'R2-Score':r2_score(actual,predicted)}
results=[]
```

```python
#step6: Base line model --> It predicts the training set for every values
baseline_predictions=np.full(len(y_test),y_train.mean())
#baseline_predictions
results.append(evaluate("Mean baseline",y_test,baseline_predictions))
results
```

```python
#step7: we apply linear regression
simple_model=LinearRegression()
#we fit the data in the model
simple_model.fit(x_train[['weight']],y_train)
simple_model_predictions=simple_model.predict(x_test[['weight']])
#print(simple_model_predictions)
results.append(evaluate("simple Linear Regression:weight",
                        y_test,
                        simple_model_predictions))
print(len(results))
```

```python
#step7: we apply linear regression
simple_model=LinearRegression()
#we fit the data in the model
simple_model.fit(x_train[['cylinders']],y_train)
simple_model_predictions=simple_model.predict(x_test[['cylinders']])
#print(simple_model_predictions)
results.append(evaluate("simple Linear Regression:cylinders",
                        y_test,
                        simple_model_predictions))
print(len(results))
```

```python
results
```

```python
#we will check the intercept value
print(f'Simple Linear Regression Equation')
print(f"Predicted MPG = {simple_model.intercept_}+({simple_model.coef_[0]}*weight")
```

```python
x['horsepower'].fillna(x['horsepower'].mean(),inplace=True)
```

```python
from sklearn.pipeline import Pipeline
```

```python
#step8: multiple linear regression
preprocessor=ColumnTransformer(transformers=[("numbers",SimpleImputer(strategy="median"),numerical_features),("category",Pipeline(steps=[("imputer",SimpleImputer(strategy="most_frequent")),("encoder",OneHotEncoder(handle_unknown="ignore",drop="first"),)]),categorical_features),])
preprocessor
```

```python
multiple_model=Pipeline(
    steps=[("prepare",preprocessor),
           ("model",LinearRegression())]
)
multiple_model.fit(x_train,y_train)
```

```python
multiple_predictions=multiple_model.predict(x_test)
results.append(evaluate("multiple_regression",y_test,multiple_predictions))
results
```

```python
#step9: polynomial Regression
#it forms a curvlinear relationship
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
polynomial_model=Pipeline(
    steps=[("imputer",SimpleImputer(strategy="median")),
           ("polynomial",PolynomialFeatures(include_bias=False)),
           ("model",LinearRegression())])
polynomial_model
```

```python
#selecting the polynomial degree
degree_search=GridSearchCV(
    polynomial_model,
    param_grid={'polynomial__degree':[1,2,3,4,5]},
    scoring="neg_mean_squared_error",
    cv=5)
```

```python
degree_search.fit(x_train[['horsepower']],y_train)
polynomial_predictions=degree_search.predict(x_test[['horsepower']])
results.append(evaluate("polynomial :Horse Power",
               y_test,polynomial_predictions))
results
```

```python
#find out the best degree (Grid search)
best_degree=degree_search.best_params_['polynomial__degree']
cv_rmse=np.sqrt(-degree_search.best_score_)
print(best_degree)
print(cv_rmse)
```

```python
#Finally we will print the results
results=pd.DataFrame(results).sort_values("RMSE")
results
```
